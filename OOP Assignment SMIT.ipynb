**Q2:** Aik `BankAccount` class banayein jismein `__balance` private variable ho. Aik `deposit()` aur aik `withdraw()` method banayein, check karein ke withdraw karte waqt balance zero se kam na ho.

`````python
class BankAccount:
    def __init__(self, initial_balance=0):
        self.__balance = initial_balance

    def deposit(self,amount):
        if amount > 0:
            self.__balance += amount
            print(f" RS {amount} has been deposited to your account. New Balance is {amount}")
        else:
            print(f"Depositing Amount Should be Greater than 0")

    def withdrawal(self,amount):
        if amount <= 0:
            print(f"Insufficient Balance!")

        elif amount > self.__balance:
            print(f"Withdrawal amount is greater than your account Balance")
        
        else:
            self.__balance = amount
            print(f"Withdrawal Successfull. Rs {amount} withdrawed Successfully. Remaining Balance{self.__balance}")


    def get_balance(self):
        return self.__balance
          
    
    
`````

**Q3:** Aik class banayein `Employee` jismein salary private ho. Agar koi salary ko minus (negative) value set karne ki koshish kare, toh setter method 'Invalid Salary' ka error message show kare.

`````python
class Employee:
    def __init__(self, name, salary):
        self.name = name
        # Constructor ke andar hi setter method ko call karne ke liye property ka istemal kiya hai
        self.salary = salary 

    # Getter method: Private salary ko access karne ke liye
    @property
    def salary(self):
        return self.__salary

    # Setter method: Salary set karte waqt negative value check karne ke liye
    @salary.setter
    def salary(self, value):
        if value < 0:
            print("Error: Invalid Salary")
            # Agar pehle se koi salary set nahi hai (constructor ke waqt), toh initial value 0 set kar dega
            if not hasattr(self, '_Employee__salary'):
                self.__salary = 0
        else:
            self.__salary = value
`````

**Q4:** Protected variable (`_`) aur Private variable (`__`) ke beech kya farq hota hai? Aik simple code ke zariye dono ka use-case dikhayein.

`````python
class Parent:
    def __init__(self):
        self._protected_var = "Main Protected Hoon"
        self.__private_var = "Main Private Hoon"

class Child(Parent):
    def display(self):
        print("Child Class se access:", self._protected_var)
        
        try:
            print(self.__private_var)
        except AttributeError:
            print("Child Class se access: NAKAM! (Private variable nahi mila)")

obj = Parent()

print(obj._protected_var)  # Access ho jayega, kyunki Python isay rokti nahi hai

# 2. Private Variable ko check karna
try:
    print(obj.__private_var)  # Error aayega
except AttributeError:
    print("Class ke bahar se access: NAKAM! (Private variable hidden hai)")

# 3. Name Mangling ke zariye Private ko access karna (Special Case)
print("Name Mangling ke sath:", obj._Parent__private_var)  # Yeh access ho jayega
`````

**Q6:** Single Inheritance ka use karte hue aik parent class `Vehicle` (with brand attribute) aur child class `Car` (with model attribute) banayein. Car ka object bana kar dono attributes print karein.

`````python
# Parent Class
class Vehicle:
    def __init__(self, brand):
        self.brand = brand  # Parent class ka attribute

# Child Class jo Vehicle ko inherit kar rahi hai
class Car(Vehicle):
    def __init__(self, brand, model):
        # super() ke zariye parent class ke constructor ko call karna
        super().__init__(brand)
        self.model = model  # Child class ka apna attribute


# ---- Car ka object banana aur attributes print karna ----

# Car class ka object banaya jismein Brand aur Model dono pass kiye
my_car = Car("Toyota", "Camry")

# Dono attributes ko print karna
print(f"Vehicle Brand: {my_car.brand}")
print(f"Car Model: {my_car.model}")
`````

**Q7:** Multilevel Inheritance ka use karein. Class `Animal` -> Class `Mammal` -> Class `Dog`. Dog class ke object se parent classes ke features ko access karein.

`````python
# Base Parent Class (Level 1)
class Animal:
    def __init__(self, name):
        self.name = name

    def eat(self):
        return f"{self.name} khana kha raha hai."

# Intermediate Child Class (Level 2) - Inherits from Animal
class Mammal(Animal):
    def __init__(self, name, has_fur=True):
        super().__init__(name)  # Animal ke constructor ko call kiya
        self.has_fur = has_fur

    def walk(self):
        return f"{self.name} chal raha hai (Mammals chal sakte hain)."

# Derived Child Class (Level 3) - Inherits from Mammal
class Dog(Mammal):
    def __init__(self, name, breed):
        super().__init__(name)  # Mammal ke constructor ko call kiya
        self.breed = breed

    def bark(self):
        return f"{self.name} bhonk raha hai! (Woof Woof)"
`````

**Q9:** Child class ke constructor (`__init__`) ke andar parent class ke constructor ko call karne ke liye `super()` ka use kaise hota hai? Aik practical example likhein.

`````python
# Parent Class
class Laptop:
    def __init__(self, brand, processor):
        self.brand = brand
        self.processor = processor
        print("Parent Class (Laptop) ka constructor chal gaya.")

    def show_specs(self):
        return f"Brand: {self.brand}, Processor: {self.processor}"

# Child Class
class GamingLaptop(Laptop):
    def __init__(self, brand, processor, gpu):
        # super() ke zariye parent class ke constructor ko call kiya aur required arguments pass kiye
        super().__init__(brand, processor)
        
        # Child class ka apna extra attribute
        self.gpu = gpu
        print("Child Class (GamingLaptop) ka constructor chal gaya.")

    # Parent ke method ko extend (modify) karna
    def show_gaming_specs(self):
        # Parent ke method ko yahan bhi super() se call kiya ja sakta hai
        basic_specs = super().show_specs()
        return f"{basic_specs}, Graphics Card: {self.gpu}"
`````

**Q10:** Hierarchical Inheritance kya hoti hai? Aik parent class `Shape` banayein aur us se do alag child classes `Circle` aur `Square` banakar samjhayein.

`````python
import math

# Parent Class
class Shape:
    def __init__(self, color):
        self.color = color  # Common feature jo sab shapes mein hota hai

    def display_color(self):
        return f"Is shape ka rang {self.color} hai."

# Child Class 1 - Inherits from Shape
class Circle(Shape):
    def __init__(self, color, radius):
        super().__init__(color)  # Parent ke constructor ko call kiya
        self.radius = radius     # Circle ka apna specific feature

    def area(self):
        # Circle ke area ka formula: πr²
        return math.pi * (self.radius ** 2)

# Child Class 2 - Inherits from Shape
class Square(Shape):
    def __init__(self, color, side):
        super().__init__(color)  # Parent ke constructor ko call kiya
        self.side = side         # Square ka apna specific feature

    def area(self):
        # Square ke area ka formula: side * side
        return self.side * self.side
`````

**Q11:** Method Overriding kya hai? Parent class `Bird` banayein jisme `fly()` method ho jo print kare 'Most birds can fly'. Child class `Ostrich` banayein jo is method ko override kare aur print kare 'Ostriches cannot fly'.

`````python
# Parent Class
class Bird:
    def fly(self):
        print("Most birds can fly.")

# Child Class jo Bird ko inherit kar rahi hai
class Ostrich(Bird):
    # Parent class ke fly() method ko override (tabdeel) kiya gaya hai
    def fly(self):
        print("Ostriches cannot fly.")


# ---- Testing Method Overriding ----

# 1. Parent class ka object banana
a_bird = Bird()
print("Bird Object:")
a_bird.fly()  # Yeh parent class ka method chalaye ga
print("-" * 30)

# 2. Child class ka object banana
an_ostrich = Ostrich()
print("Ostrich Object:")
an_ostrich.fly()  # Yeh child class ka overridden method chalaye ga
`````

**Q13:** Operator Overloading kya hoti hai? Do objects ko `+` operator ke zariye add karne ke liye `__add__` magic method ka use kaise karte hain? Code likhein.

`````python
class Book:
    def __init__(self, title, price):
        self.title = title
        self.price = price

    # '+' operator ko overload karne ke liye magic method
    def __add__(self, other):
        # self -> pehla object (book1)
        # other -> doosra object (book2)
        total_price = self.price + other.price
        return total_price

    # Object ko readable format mein print karne ke liye (Optional magic method)
    def __str__(self):
        return f"'{self.title}' (Rs. {self.price})"


# ---- Testing Operator Overloading ----

# Do alag alag books ke objects banaye
book1 = Book("Python Basics", 1200)
book2 = Book("Data Structures", 1500)

print(f"Kitab 1: {book1}")
print(f"Kitab 2: {book2}")
print("-" * 40)

# Operator Overloading ka jadhoo: Do objects ko '+' se add karna
# Yeh line back-end par book1.__add__(book2) ko call karegi
kul_kharcha = book1 + book2

print(f"Dono kitabon ki kul keemat: Rs. {kul_kharcha}")
`````

**Q15:** Kya Python default roop se Method Overloading (same method name with different arguments) support karta hai? Agar nahi, toh hum same name ke method ko multiple arguments ke sath kaise handle karte hain? Code likhein.

`````python
class Calculator:
    # Default arguments (None) ka use karke overloading achieve karna
    def add(self, a, b, c = None):
        if c is not None:
            return a + b + c
        else:
            return a + b

# ---- Testing ----
calc = Calculator()

print("Do numbers ka sum:", calc.add(10, 20))      # Output: 30
print("Teen numbers ka sum:", calc.add(10, 20, 30)) # Output: 60
`````

**Q18:** Aik abstract class `Car` banayein jisme aik concrete (normal) method `start_engine()` ho aur aik abstract method `drive()` ho. Child class bana kar dono ko test karein.

`````python
from abc import ABC, abstractmethod

class Car(ABC):
    # Concrete Method (Normal Method)
    def start_engine(self):
        return "Engine start ho gaya... Vroom Vroom!"

    # Abstract Method
    @abstractmethod
    def drive(self):
        pass

# Child Class
class Sedan(Car):
    def drive(self):
        return "Sedan smooth tareeqe se road par chal rahi hai."

# ---- Testing ----
my_car = Sedan()

print(my_car.start_engine()) # Parent ka concrete method chal gaya
print(my_car.drive())        # Child ka implemented method chal gaya
`````

**Q20:** Aik real-world scenario (jaise Payment System) design karein jahan `Payment` aik abstract class ho aur `CreditCardPayment` aur `UPIPayment` uski subclasses hon.

`````python
from abc import ABC, abstractmethod

# Abstract Base Class
class Payment(ABC):
    def __init__(self, amount):
        self.amount = amount

    # Concrete Method: Jo sab payment systems mein common hai
    def transaction_status(self):
        print(f"Processing transaction of Rs. {self.amount}...")

    # Abstract Method: Har payment system ka apna security process hota hai
    @abstractmethod
    def validate_and_pay(self):
        pass

# Subclass 1: Credit Card Payment
class CreditCardPayment(Payment):
    def __init__(self, amount, card_number):
        super().__init__(amount)
        self.card_number = card_number

    def validate_and_pay(self):
        self.transaction_status()
        print(f"OTP sent to card ending in {self.card_number[-4:]}.")
        print("Credit Card Payment Successful!\n")

# Subclass 2: UPI Payment
class UPIPayment(Payment):
    def __init__(self, amount, upi_id):
        super().__init__(amount)
        self.upi_id = upi_id

    def validate_and_pay(self):
        self.transaction_status()
        print(f"Verifying UPI PIN for ID: {self.upi_id}")
        print("UPI Payment Successful!\n")


# ---- Payment System Testing ----

# 1. Credit Card ke zariye payment
card_pay = CreditCardPayment(5000, "1234-5678-9876-5432")
card_pay.validate_and_pay()

# 2. UPI ke zariye payment
upi_pay = UPIPayment(1200, "user@okaxis")
upi_pay.validate_and_pay()
`````
